```python
# with args 

def my_decorator(func):
    def wrapper(*args, ):
        print("Before")
        result = func(*args)
        print("After")
        return result
    return wrapper
 
@my_decorator
def add(a, b):
    return a + b

print(add(3, 4))
        
```

```python
nums = [1,2,3,4,5,6,7]  
def get_number():
    for i in nums:
        yield i 

number = get_number()
print(number)

print(next(number))
```
