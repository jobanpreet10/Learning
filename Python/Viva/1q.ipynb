```python
lis = [1,3,2,5,4]
target = 8
# output = 1 3
```

```python
def twoSum(lis,target):
    n = len(lis)
    dic = {}
    for k  in range(n): 
        comp = target - lis[k]
        if comp in dic:
            return [dic[comp],k]
        
        dic[lis[k]] = k

    return []


{7 : 0 , 5 : 1, 6: 2}
```
